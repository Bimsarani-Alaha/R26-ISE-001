````python
import os
import subprocess
import time

print("Starting Ollama server...")
subprocess.Popen(
    ["ollama", "serve"],
    env={**os.environ, "OLLAMA_HOST": "0.0.0.0:11434"},
    stdout=open("/tmp/ollama.log", "w"),
    stderr=subprocess.STDOUT,
)
for _ in range(60):
    try:
        subprocess.run(
            ["curl", "-s", "-f", "http://localhost:11434/api/tags"],
            capture_output=True,
            timeout=5,
        )
        print("Ollama is up!")
        break
    except Exception:
        time.sleep(2)
````

````python
from pyngrok import ngrok
import re

# Get your free auth token from https://dashboard.ngrok.com/get-started/your-authtoken
NGROK_AUTH_TOKEN = ""  # <-- PASTE YOUR NGROK AUTH TOKEN HERE

if not NGROK_AUTH_TOKEN:
    raise ValueError(
        "Set NGROK_AUTH_TOKEN above.\n"
        "1. Sign up at https://ngrok.com (free)\n"
        "2. Go to https://dashboard.ngrok.com/get-started/your-authtoken\n"
        "3. Copy the token and paste it above"
    )

ngrok.set_auth_token(NGROK_AUTH_TOKEN)
public_url = ngrok.connect(11434)
print(f"\n{'='*50}")
print(f"OLLAMA_BASE_URL={public_url}")
print(f"{'='*50}")
print("\nCopy the URL above into backend/.env and restart run_backend.bat")
````
